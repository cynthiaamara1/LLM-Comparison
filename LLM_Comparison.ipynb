```python
# List of the AI models to be tested
models = [
    "qwen/qwen3-32b",
    "deepseek/deepseek-v4.1-flash",
    "meta-llama/llama-3.3-70b-instruct",
    "google/gemma-3-27b-it",
    "mistralai/ministral-3b-2512"
]

# Store the summaries generated by each model
all_summaries = {}

# Store the results from the different tests
all_results = {}
```

```python
import time
```

```python
# Function to test different models using the same prompts
def evaluate_models(models, system_prompt, user_prompt):

    # Dictionary to store the results from each model
    results = {}

    # Go through each model one by one
    for model in models:

        # Start the timer to measure how long the model takes
        start = time.time()

        # Send the prompts to the selected model
        response = client.chat.completions.create(
            model=model,
            messages=[
                {
                    # Tell the model how to respond
                    "role": "system",
                    "content": system_prompt
                },
                {
                    # Provide the task or question to the model
                    "role": "user",
                    "content": user_prompt
                }
            ],
            max_tokens=4000  # Set the maximum number of tokens for the response
        )

        # Calculate the time taken to generate the response
        elapsed = time.time() - start

        # Save the response, time taken and token usage
        results[model] = {
            "response": response.choices[0].message.content,
            "latency": elapsed,
            "prompt_tokens": response.usage.prompt_tokens,
            "completion_tokens": response.usage.completion_tokens,
            "total_tokens": response.usage.total_tokens
        }

    # Return all the results after testing the models
    return results
```

```python
# Function to display the results from each model
def print_results(results):

    # Go through each model and its results
    for model, result in results.items():

        # Print a line to separate each model's results
        print("=" * 80)

        # Display the model name
        print(f"Model: {model}")

        # Display the time taken by the model
        print(f"Latency: {result['latency']:.2f} seconds")

        # Display the number of tokens used in the prompt
        print(f"Prompt Tokens: {result['prompt_tokens']}")

        # Display the number of tokens generated in the response
        print(f"Completion Tokens: {result['completion_tokens']}")

        # Print a line before displaying the response
        print("-" * 80)

        # Display the model's response
        print(result["response"])

        # Add an empty line before the next model's results
        print()
```

#Text Summarization

```python
# Set the instruction that tells the models what to do
system_prompt = (
    """You are a top 1% story teller and reviewer. Summarize story in 5 bullet points."""
)

# Store the story that will be sent to the models
prompt = story

# Run the evaluation using all selected models and the same prompts
Review_generator = evaluate_models(
    models,
    system_prompt,
    prompt
)

# Display the responses and performance results from each model
print_results(Review_generator)
```
